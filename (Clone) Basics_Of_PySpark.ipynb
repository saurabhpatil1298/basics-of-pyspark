```python
%sql
use my_catalog.my_schema;
create volume my_volume; -- To upload the files
```

**Create data frame using csv file ⬇️**

```python
# The default value for the header and inferShema is false. So, the hearder will provide the values like c0, c1, c2..etc. and the inferSchema will not infer the schema and will consider the data type as string.
# If the header is false and inferSchema is true, then the header will provide the values like c0, c1, c2..etc. and the inferSchema will infer the schema but will consider the data type as string only as the header is false.
# If the header is true and inferSchema is false, then the header will provide the actule values and the inferSchema will not infer the schema and will consider the data type as string.
# If the header is true and inferSchema is true, then the header will provide the actule values and the inferSchema will infer the schema and will consider the data type as per the data.

path = r'/Volumes/my_catalog/my_schema/my_volume/csv_files/employees.csv'
csv_df = spark.read.csv(path=path, header=True, inferSchema=True)
display(csv_df)
csv_df.printSchema() # Prints the schema of the dataframe
```

**Create data frame using parquet file ⬇️**

```python
path1 = r'/Volumes/my_catalog/my_schema/my_volume/parquet_files/part-00000-a9e77425-5fb4-456f-ba52-f821123bd193-c000.snappy.parquet'
parquet_df = spark.read.parquet(path1)
display(parquet_df)
parquet_df.printSchema()
```

**Create data frame using json file ⬇️**

```python
path2 = r'/Volumes/my_catalog/my_schema/my_volume/json_files/events.json'
json_df = spark.read.json(path2, multiLine=True)
display(json_df)
json_df.printSchema()
```

**Reading files using option() method ⬇️**

```python
df_csv = spark.read.option('header', True).option('inferSchema', True).csv(path)
display(df_csv)

df_parquet = spark.read.option('header', True).option('inferSchema', True).parquet(path1)
display(df_parquet)

df_json = spark.read.option('multiLine', True).json(path2)
display(df_json)
```

```python
# Print column names from dataframe
print(df_csv.columns)
print(df_parquet.columns)
print(df_json.columns)
```

```python
# Print count of columns from dataframe
print('The count of column in the df_csv dataframe is: ',len(df_csv.columns))
print('The count of column in the df_parquet dataframe is: ',len(df_parquet.columns))
print('The count of column in the df_json dataframe is: ',len(df_json.columns))
```

```python
# Print column names and theire data types from dataframe
print('Column names and theire data types from df_csv dataframe are:')
for df in df_csv.dtypes:
    print(df)
print('\nColumn names and theire data types from df_csv dataframe are:')
for df in df_parquet.dtypes:
    print(df)
print('\nColumn names and theire data types from df_csv dataframe are:')
for df in df_json.dtypes:
    print(df)
```

```python
# Create list of column names having same data type from dataframe
int_data_type_columns = []
string_data_type_columns = []
other_data_type_columns = []

for df in df_csv.dtypes:
    if df[1]=='int':
        int_data_type_columns.append(df[0])
    elif df[1]=='string':
        string_data_type_columns.append(df[0])
    else:
        other_data_type_columns.append(df[0])
print('The list of column names having int data type from df_csv dataframe are: ',int_data_type_columns)
print('The list of column names having string data type from df_csv dataframe are: ',string_data_type_columns)
print('The list of column names having other data type from df_csv dataframe are: ',other_data_type_columns)

for df in df_parquet.dtypes:
    if df[1]=='int':
        int_data_type_columns.append(df[0])
    elif df[1]=='string':
        string_data_type_columns.append(df[0])
    else:
        other_data_type_columns.append(df[0])
print('\nThe list of column names having int data type from df_parquet dataframe are: ',int_data_type_columns)
print('The list of column names having string data type from df_parquet dataframe are: ',string_data_type_columns)
print('The list of column names having other data type from df_parquet dataframe are: ',other_data_type_columns)

for df in df_json.dtypes:
    if df[1]=='int':
        int_data_type_columns.append(df[0])
    elif df[1]=='string':
        string_data_type_columns.append(df[0])
    else:
        other_data_type_columns.append(df[0])
print('\nThe list of column names having int data type from df_json dataframe are: ',int_data_type_columns)
print('The list of column names having string data type from df_json dataframe are: ',string_data_type_columns)
print('The list of column names having other data type from df_json dataframe are: ',other_data_type_columns)
```
